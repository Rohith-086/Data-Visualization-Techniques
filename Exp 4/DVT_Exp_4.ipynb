```python
import kagglehub
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# ============================================================
# LOAD DATASET FROM KAGGLE
# ============================================================

path = kagglehub.dataset_download("jessemostipak/hotel-booking-demand")

print("Dataset Path:")
print(path)

# Find CSV file automatically
csv_files = [f for f in os.listdir(path) if f.endswith(".csv")]

print("\nCSV Files Found:")
print(csv_files)

csv_file = csv_files[0]

# Load Dataset
df = pd.read_csv(
    os.path.join(path, csv_file),
    encoding="latin1",
    low_memory=False
)

print("\nDataset Loaded Successfully")

# ============================================================
# PROGRAM 1 - DATA INSPECTION
# ============================================================

print("\nFirst 5 Records")
print(df.head())

print("\nLast 5 Records")
print(df.tail())

# ============================================================
# PROGRAM 2 - SHAPE
# ============================================================

print("\nShape of Dataset")
print(df.shape)

print("\nRows =", df.shape[0])
print("Columns =", df.shape[1])

# ============================================================
# PROGRAM 3 - DATA TYPES
# ============================================================

print("\nData Types")
print(df.dtypes)

print("\nDataset Info")
df.info()

# ============================================================
# PROGRAM 4 - FILTERING
# Bookings with Lead Time > 100 Days
# ============================================================

long_lead = df[df["lead_time"] > 100]

print("\nBookings with Lead Time Greater Than 100 Days")

print(
    long_lead[
        [
            "hotel",
            "is_canceled",
            "lead_time",
            "arrival_date_year",
            "market_segment",
            "customer_type"
        ]
    ].head(10)
)

# ============================================================
# PROGRAM 5 - FILTER CANCELED BOOKINGS
# ============================================================

canceled = df[df["is_canceled"] == 1]

print("\nCanceled Bookings")

print(
    canceled[
        [
            "hotel",
            "is_canceled",
            "lead_time",
            "market_segment",
            "customer_type"
        ]
    ].head(10)
)

# ============================================================
# PROGRAM 6 - SORTING
# Sort by Average Daily Rate
# ============================================================

sorted_df = df.sort_values(
    by="adr",
    ascending=False
)

print("\nTop Bookings by Average Daily Rate")

print(
    sorted_df[
        [
            "hotel",
            "arrival_date_year",
            "market_segment",
            "customer_type",
            "adr",
            "is_canceled"
        ]
    ].head(10)
)

# ============================================================
# BAR CHART
# TOP 10 BOOKINGS BY ADR
# ============================================================

top10 = sorted_df.head(10)

plt.figure(figsize=(10, 5))

plt.bar(
    range(len(top10)),
    top10["adr"]
)

plt.xticks(
    range(len(top10)),
    top10["hotel"],
    rotation=90
)

plt.xlabel("Hotel")
plt.ylabel("Average Daily Rate (ADR)")
plt.title("Top 10 Hotel Bookings by ADR")

plt.show()

# ============================================================
# PROGRAM 7 - STATISTICAL ANALYSIS
# ============================================================

print("\nMean ADR =")
print(df["adr"].mean())

print("\nMedian ADR =")
print(df["adr"].median())

print("\nMode ADR =")
print(df["adr"].mode())

print("\nStandard Deviation =")
print(df["adr"].std())

print("\nMinimum ADR =")
print(df["adr"].min())

print("\nMaximum ADR =")
print(df["adr"].max())

# ============================================================
# PROGRAM 8 - DESCRIPTIVE STATISTICS
# ============================================================

print("\nDescriptive Statistics")

print(
    df[
        [
            "lead_time",
            "stays_in_weekend_nights",
            "stays_in_week_nights",
            "adults",
            "children",
            "babies",
            "adr"
        ]
    ].describe()
)

# ============================================================
# PROGRAM 9 - GROUPING
# Average ADR by Hotel Type
# ============================================================

group = df.groupby(
    "hotel"
)["adr"].mean().sort_values(ascending=False)

print("\nAverage ADR by Hotel Type")
print(group)

# Visualization

group.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.xlabel("Hotel Type")
plt.ylabel("Average ADR")
plt.title("Average ADR by Hotel Type")

plt.xticks(rotation=0)

plt.show()

# ============================================================
# PROGRAM 10 - AGGREGATION
# ============================================================

summary = df.groupby("hotel").agg({
    "adr": "mean",
    "lead_time": "mean",
    "is_canceled": "mean"
})

print("\nHotel Aggregation")
print(summary)

# ============================================================
# PROGRAM 11 - CORRELATION
# ============================================================

numeric = df.select_dtypes(
    include=np.number
)

corr = numeric.corr()

print("\nCorrelation Matrix")
print(corr)

plt.figure(figsize=(12, 8))

sns.heatmap(
    corr,
    annot=False,
    cmap="coolwarm"
)

plt.title("Hotel Booking Dataset Correlation Heatmap")

plt.show()

# ============================================================
# PROGRAM 12 - SCATTER PLOT
# Lead Time vs ADR
# ============================================================

plt.figure(figsize=(7, 5))

plt.scatter(
    df["lead_time"],
    df["adr"]
)

plt.xlabel("Lead Time")
plt.ylabel("Average Daily Rate (ADR)")
plt.title("Lead Time vs Average Daily Rate")

plt.show()

# ============================================================
# PROGRAM 13 - HISTOGRAM
# ADR Distribution
# ============================================================

plt.figure(figsize=(7, 5))

plt.hist(
    df["adr"].dropna(),
    bins=30
)

plt.xlabel("Average Daily Rate (ADR)")
plt.ylabel("Frequency")
plt.title("Distribution of Hotel Booking ADR")

plt.show()

# ============================================================
# PROGRAM 14 - BOXPLOT
# ============================================================

plt.figure(figsize=(7, 5))

sns.boxplot(
    x=df["adr"].dropna()
)

plt.title("ADR Boxplot")
plt.xlabel("Average Daily Rate (ADR)")

plt.show()

# ============================================================
# PROGRAM 15 - VALUE COUNTS
# ============================================================

print("\nMarket Segment Counts")

print(
    df["market_segment"].value_counts()
)

# ============================================================
# PROGRAM 16 - UNIQUE VALUES
# ============================================================

print("\nUnique Hotels")

print(
    df["hotel"].unique()
)

print("\nUnique Market Segments")

print(
    df["market_segment"].unique()
)

print("\nUnique Customer Types")

print(
    df["customer_type"].unique()
)

# ============================================================
# PROGRAM 17 - TOP 10 COUNTRIES BY BOOKINGS
# ============================================================

print("\nTop 10 Countries by Number of Bookings")

country_counts = (
    df["country"]
    .value_counts()
    .head(10)
)

print(country_counts)

# ============================================================
# PROGRAM 18 - BOOKINGS BY YEAR
# ============================================================

print("\nNumber of Bookings by Year")

year_counts = (
    df["arrival_date_year"]
    .value_counts()
    .sort_index()
)

print(year_counts)

# ============================================================
# PROGRAM 19 - TOP 10 COUNTRIES BY CANCELLATIONS
# ============================================================

print("\nTop 10 Countries by Canceled Bookings")

country_cancellations = (
    df[df["is_canceled"] == 1]
    .groupby("country")
    .size()
    .sort_values(ascending=False)
    .head(10)
)

print(country_cancellations)

# ============================================================
# PROGRAM 20 - MOST COMMON MARKET SEGMENTS
# ============================================================

print("\nMost Common Market Segments")

market_segments = (
    df["market_segment"]
    .value_counts()
    .head(10)
)

print(market_segments)

# ============================================================
# PROGRAM 21 - EXPORT DATA
# ============================================================

df.to_csv(
    "hotel_booking_data.csv",
    index=False
)

df.to_excel(
    "hotel_booking_data.xlsx",
    index=False
)

df.to_json(
    "hotel_booking_data.json",
    orient="records",
    indent=4
)

print("\nCSV, Excel and JSON files exported successfully.")

# ============================================================
# FINAL REPORT
# ============================================================

print("\n==============================================")
print("EXPERIMENT COMPLETED SUCCESSFULLY")
print("==============================================")

print("Dataset Shape :", df.shape)

print(
    "Missing Values :",
    df.isnull().sum().sum()
)

print(
    "Total Bookings :",
    len(df)
)

print(
    "Total Canceled Bookings :",
    df["is_canceled"].sum()
)

print(
    "Average ADR :",
    df["adr"].mean()
)

print(
    "Number of Countries :",
    df["country"].nunique()
)

print(
    "Number of Market Segments :",
    df["market_segment"].nunique()
)

print(
    "Number of Customer Types :",
    df["customer_type"].nunique()
)

print("Correlation Matrix Generated")

print("==============================================")
```
