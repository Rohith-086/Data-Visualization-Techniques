```python
import pandas as pd
import os
import kagglehub
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("hotel_booking.csv")
```

```python

# --------------------------------
# 1. Line Chart – Bookings by Year
# --------------------------------

yearly_bookings = df.groupby("arrival_date_year").size()

plt.figure(figsize=(10, 5))

plt.plot(
    yearly_bookings.index,
    yearly_bookings.values
)

plt.title("Hotel Bookings by Year")
plt.xlabel("Arrival Year")
plt.ylabel("Number of Bookings")
plt.grid(True)

plt.show()


# --------------------------------
# 2. Bar Chart – Bookings by Hotel Type
# --------------------------------

hotel_bookings = df["hotel"].value_counts()

plt.figure(figsize=(10, 5))

plt.bar(
    hotel_bookings.index,
    hotel_bookings.values
)

plt.title("Hotel Bookings by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Number of Bookings")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()

plt.show()


# --------------------------------
# 3. Pie Chart – Booking Distribution
# --------------------------------

top_segments = df["market_segment"].value_counts().head(6)

plt.figure(figsize=(8, 8))

plt.pie(
    top_segments.values,
    labels=top_segments.index,
    autopct="%1.1f%%"
)

plt.title("Top 6 Market Segments by Hotel Bookings")

plt.show()
```
