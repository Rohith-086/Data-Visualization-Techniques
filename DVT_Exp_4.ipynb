```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("hotel_booking.csv")
```

```python
print(df.head())

print(df.shape)

print(df.dtypes)
```

```python
print(df.describe())
```

```python
high_adr = df[df["adr"] > 200]
print(high_adr.head())
```

```python
city_hotel = df[df["hotel"] == "City Hotel"]

print(city_hotel.head())
```

```python
sorted_df = df.sort_values(
    by="adr",
    ascending=False
)

print(sorted_df.head())
```

```python
print("Mean :", df["adr"].mean())

print("Median :", df["adr"].median())

print("Mode :")

print(df["adr"].mode())

print("Standard Deviation :", df["adr"].std())
```

```python
hotel_avg = df.groupby("hotel")["adr"].mean()

print(hotel_avg.sort_values(ascending=False))
```

```python
hotel_avg.sort_values(ascending=False).plot(kind="bar")

plt.title("Average ADR by Hotel Type")

plt.ylabel("Average ADR")

plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="lead_time",
    y="adr"
)

plt.title("Lead Time vs ADR")

plt.xlabel("Lead Time")

plt.ylabel("ADR")

plt.show()
```

```python
corr = df.corr(numeric_only=True)

plt.figure(figsize=(10,8))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.hist(
    df["adr"],
    bins=30
)

plt.title("ADR Distribution")

plt.xlabel("ADR")

plt.ylabel("Frequency")

plt.show()
```
